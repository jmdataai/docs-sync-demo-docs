# Top Cities chart: Gold query

The SQL the chart runs through the Tenant API. The next cell is written by docs sync.

```sql
SELECT s.city_nm, SUM(s.net_revenue_amt) AS revenue
FROM F_SALES_SUMMARY s
WHERE s.sale_dt BETWEEN :startDate
AND :endDate
AND s.region_cd IN (SELECT r.region_cd
FROM D_REGION r
WHERE r.active_flg = 'Y')
AND s.region_cd <> 'TEST'
GROUP BY s.city_nm
ORDER BY revenue DESC
LIMIT 10
```

Run it in Snowflake with real dates in place of `:startDate` and `:endDate`.

````python
%sql
SELECT s.city_nm, SUM(s.net_revenue_amt) AS revenue
FROM F_SALES_SUMMARY s
WHERE s.sale_dt BETWEEN '2026-01-01' AND '2026-03-31'
GROUP BY s.city_nm ORDER BY revenue DESC LIMIT 10
````
